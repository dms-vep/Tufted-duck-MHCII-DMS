```python
import pandas as pd
from pathlib import Path

frags = pd.read_csv("tufted_duck_MHCII_assembly_fragments.csv")
sites = sorted({i for a,b in frags[["start_site","end_site"]].itertuples(index=False)
                for i in range(int(a), int(b)+1)})

seq = "".join(l.strip() for l in Path("tufted_duck_MHCII.fa").read_text().splitlines()
              if l and not l.startswith(">"))

aas = list("ACDEFGHIKLMNPQRSTVWY")
first20 = set(sites[:20])

rows = []
for s in sites:
    wt = seq[s-1]
    muts = aas + (["*"] if s in first20 else [])
    rows.extend((s, wt, m, 1) for m in muts if m != wt)

df = pd.DataFrame(rows, columns=['sequential_site', 'wildtype_aa', 'mutant_aa', 'representation'])
df.to_csv("mutations_to_make.csv", index=False)
df.head()
```
